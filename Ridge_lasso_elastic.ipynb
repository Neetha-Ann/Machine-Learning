```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.model_selection import GridSearchCV

from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import ElasticNet

import joblib
```

EDA

```python
data = pd.read_csv('Data/Advertising.csv')
data.head()
```

```python
print(data.columns.tolist())
```

```python
data.drop(columns='Unnamed: 0', inplace=True)
```

```python
print(data.columns.tolist())
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
x = data.drop(['Sales'],axis =1)
y = data['Sales']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=.33, random_state=42)
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)
y_pred
```

MSE-Testing error

```python
lr_testing_error = mean_squared_error(y_pred, y_test)
lr_testing_error
```

Training error

```python
ypred_train = lr.predict(x_train)
training_error = mean_squared_error(ypred_train,y_train)
training_error
```

```python
print("Train R2:", r2_score(y_train, ypred_train))
print("Test R2:", r2_score(y_test, y_pred))
```

Ridge

```python
alphavalues = [1e-15, 1e-10, 1e-8, 1e-4, 1e-3, 1e-2, 1, 5, 10, 20, 30, 100]

solvervalues=['auto', 'svd']

r = Ridge()

parameter={'alpha': alphavalues, 'solver': solvervalues}
ridge_regressor=GridSearchCV(r, parameter)
ridge_regressor.fit(x_train, y_train)
```

```python
ridge_regressor.best_estimator_
```

```python
ridge_regressor.best_params_
```

```python
ridge_regressor.score(x_train, y_train)
```

```python
ridge=Ridge(alpha=100, solver='auto')
ridge.fit(x_train, y_train)
y_pred_ridge=ridge.predict(x_test)
y_pred_ridge
```

```python
te = mean_squared_error(y_pred_ridge, y_test)
te
```

```python
ridge.score(x_train,y_train)
```

Lasso

```python
la = Lasso()
parameters={'alpha': [1e-15, 1e-10, 1e-8, 1e-4, 1e-3, 1e-2, 1, 5, 10, 20]}
lasso_regresssor = GridSearchCV(la, parameters)
lasso_regresssor.fit(x_train, y_train)
```

```python
bp = lasso_regresssor.best_params_
bp
```

```python
lasso=Lasso(alpha=bp['alpha'])
lasso.fit(x_train,y_train)
ypl=lasso.predict(x_test)
ypl
```

```python
lte = mean_squared_error(ypl, y_test)
lte
```

```python
lasso.score(x_train, y_train)
```

Elastic Net

```python
elastic=ElasticNet()
parameters = {'alpha': [1e-15, 1e-10, 1e-8, 1e-4, 1e-3, 1e-2, 1, 5, 10, 20]}

elastic_regressor = GridSearchCV(elastic, parameters)
elastic_regressor.fit(x_train, y_train)
```

```python
elastic_regressor.best_params_
```

```python
elastic=ElasticNet(alpha=1)
elastic.fit(x_train, y_train)
ype = elastic.predict(x_test)
ype
```

```python
ete = mean_squared_error(ype,y_test)
ete
```

```python
elastic.score(x_train,y_train)
```

```python
print('Ridge:', te)
print('Lasso:', lte)
print('elastic:', ete)
```

How to save a model

```python
joblib.dump(elastic,'Adv.pkl' )
```

Load the saved model

```python
model_from_dir = joblib.load('Adv.pkl')
ypred = model_from_dir.predict(x_test)
ypred
```
