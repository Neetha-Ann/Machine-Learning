```python
y=data['diagnosis']
x=data.drop(['id', 'diagnosis'], axis =1)
```

```python
x_train, x_test, y_train, y_test =train_test_split(x,y,test_size=0.33, random_state=42, stratify=y)
```

```python
preprocessor = ColumnTransformer([
  ('num', Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
  ]),numeric_features),
])

print('Train:', x_train.shape, '| Test:', x_test.shape)
print('classes:', y_train.unique().tolist())

```

```python
lg = LogisticRegression(max_iter=1000, random_state=42)
print('preprocessed x train values:', x_train)
lg.fit(x_train,y_train)
```

```python
y_pred = lg.predict(x_test)
y_pred 
```

```python
y_pred = lg.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

```python
#confusion matrix
confusion_matrix(y_test, y_pred)
```

Using for loop try confusion matrix

```python
tp = 0 #True Positives
tn = 0 #True Negatives
fp = 0 #False Positives
fn = 0 #False Negatives

y_test_list = y_test.tolist()

#Assuming y_true and y_pred are lists or arrays containing the actual and predicted labels
for i in range(len(y_pred)):
  if y_pred[i] ==1: #Predicted Positive
    if y_test_list[i] ==1: # True Positive
      tp +=1
    else: #False Positive
      if y_test_list[i] ==1: #False Negative
        fn+= 1
      else: #True Negaive
        tn+=1
        
# Output the confusion matrix
print(f'True Positive (TP): {tp}')
print(f'True Negative (TN): {tn}')
print(f'False Positive (FP): {fp}')
print(f'False Negative (FN): {fn}')
    

```
